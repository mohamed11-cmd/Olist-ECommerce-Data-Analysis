# 4.EDA




## a.Overall business

```python
####Extract 'year' from 'order_purchase_timestamp'
orders['year']=orders['order_purchase_timestamp'].dt.to_period("Y")
```

```python
## Number of orders per month among the 3 years
orders_year=orders.groupby('year')['order_id'].nunique().nlargest(3).reset_index()
orders_year
```

```python
##Extract year_month from 'order_purchase_timestamp'
orders['year_month']=orders['order_purchase_timestamp'].dt.to_period("M")
```

```python
## Number of orders per month among the 3 years
orders_year_month=orders.groupby('year_month')['order_id'].nunique().nlargest(5).reset_index()
orders_year_month
```

```python
## Merge orders table & order_items table to see sales flow
orders_revenue = orders.merge(order_items, on='order_id', how='inner')
```

```python
##sales flow on time
orders_revenue['year']=orders_revenue['order_purchase_timestamp'].dt.year
sales_year=orders_revenue.groupby('year')['price'].sum().sort_index(ascending=False).reset_index()
sales_year
```

```python
##sales flow on year_month
orders_revenue['year_month']=orders_revenue['order_purchase_timestamp'].dt.to_period("M")
sales_year_m=orders_revenue.groupby('year_month')['price'].sum().nlargest().reset_index()
sales_year_m
```

## e.Logistics 

```python
## compare between delay orders & no delay orders
delivered_orders['is_delay']=(delivered_orders['order_delivered_customer_date']>delivered_orders['order_estimated_delivery_date'])
delivered_orders['is_delay'].value_counts()
```

```python
## Merge delivered_orders & customers
cust_deliv=delivered_orders.merge(customers,on='customer_id',how='inner')
```

```python
## Calculate diff between 'order_delivered_customer_date' & 'order_estimated_delivery_date'
cust_deliv['delay_days']=(cust_deliv['order_delivered_customer_date']-cust_deliv['order_estimated_delivery_date']).dt.days
cust_deliv['delay_days'].mean()
```

```python
## cities that orders arrive before estimated_delivery_date
customer_delay_L=cust_deliv.groupby('customer_city')['delay_days'].mean().nsmallest().reset_index()
customer_delay_L
```

```python
## cities that orders arrive after estimated_delivery_date
customer_delay_s=cust_deliv.groupby('customer_city')['delay_days'].mean().nlargest(10).reset_index()
customer_delay_s
```

```python
## Days that orders take to arrive from purchase_timestamp untill delivered_date
cust_deliv['order_trip']=(cust_deliv['order_delivered_customer_date']-cust_deliv['order_purchase_timestamp']).dt.days
```

```python
## orders take a long trip
dliv_days_L=cust_deliv.groupby('customer_city')['order_trip'].mean().nlargest(10).reset_index()
dliv_days_L
```

```python
## orders take short trip 
dliv_days_s=cust_deliv.groupby('customer_city')['order_trip'].mean().nsmallest(10).reset_index()
dliv_days_s
```

## f.Reviews

```python
## Compare between scores
order_review[['review_score']].value_counts().nlargest(5)
```

```python
## Merge order_review table & cust_deliv table
review_delay=order_review.merge(cust_deliv,on='order_id',how='inner')
```

```python
# بنحدد الحدود الفاصلة بين الصناديق
bins = [-float('inf'), -7, -1, 0, 7, float('inf')]
# بنسمي الصناديق دي عشان تبان في الرسمة
labels = ['Early (>7d)', 'Early (1-7d)', 'On Time', 'Late (1-7d)', 'Late (>7d)']
# بنقول لبايثون: خُد عمود delay_days وقسّمه بناءً على الحدود دي
review_delay['delay_category'] = pd.cut(review_delay['delay_days'], bins=bins, labels=labels)
```

```python
## Rating orders that arrive before their time 
customers_review=review_delay.groupby('delay_category')['review_score'].mean().reset_index()
customers_review
```

# 7.Business Problems & Recommendations



### 1. Delivery Delays & Customer Satisfaction
* **Problem:** Delivery delays are directly correlated with lower customer review scores, which damages platform reputation and customer retention.
* **Recommendation:** 
  * Implement Service Level Agreements (SLAs) with strict dispatch deadlines for sellers.
  * Establish automated tracking and alert systems to proactively notify customers of potential shipping delays.
  * Optimize logistics and fulfillment nodes in regions outside São Paulo to reduce delivery lead times.

### 2. Market Concentration Risk
* **Problem:** Over-reliance on São Paulo as the primary driver for both customer purchasing power and seller locations creates geographical risk and higher shipping costs for remote regions.
* **Recommendation:**
  * Incentivize seller onboarding and regional fulfillment centers in underserved states to decentralize supply.
  * Run targeted marketing campaigns in secondary high-potential markets to expand the active customer base.

### 3. Seasonality & Inventory Management
* **Problem:** Demand peaks sharply in specific periods (e.g., November / Black Friday), risking stockouts or supply chain bottlenecks for high-revenue categories like **Health & Beauty**.
* **Recommendation:**
  * Implement predictive sales forecasting to help top sellers manage inventory levels ahead of Q4 demand spikes.
  * Create promotional product bundles combining high-demand items (Health & Beauty) with slower-moving inventory to boost Average Order Value (AOV).

### 4. Payment & Checkout Optimization
* **Problem:** High-value purchases depend heavily on multi-installment credit card plans; friction in payment options directly impacts conversion rates for expensive items.
* **Recommendation:**
  * Partner with financial institutions to offer zero-interest installment plans for high-ticket product categories.
  * Enhance checkout experience by featuring preferred payment options (Credit Card & Installments) prominently.
